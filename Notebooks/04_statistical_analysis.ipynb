# Step 4 — Statistical Analysis

This notebook statistically investigates relationships and group differences in the Student Performance dataset using **SciPy**.

### Tests included
- Pearson correlation: Study Time vs Final Grade
- Pearson correlation: Absences vs Final Grade
- Pearson correlation: Previous Failures vs Final Grade
- Independent Welch's t-test: Gender vs Final Grade
- Independent Welch's t-test: Internet Access vs Final Grade
- Independent Welch's t-test: School vs Final Grade
- One-way ANOVA: Study Time vs Final Grade
- G1/G2 vs G3 correlation analysis
- Automatic statistical summary table

**Significance level:** α = 0.05


```python
import pandas as pd
import numpy as np
from scipy import stats

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: f'{x:.4f}')

ALPHA = 0.05

print('Libraries imported successfully!')
```

```python
# Load cleaned dataset
df = pd.read_csv('../data/student_performance_cleaned.csv')

print('Dataset shape:', df.shape)
display(df.head())
```

## 1. Study Time vs Final Grade — Pearson Correlation

**H₀:** There is no linear correlation between study time and final grade.

**H₁:** There is a linear correlation between study time and final grade.

```python
study_corr, study_p = stats.pearsonr(df['studytime'], df['G3'])

print(f'Correlation: {study_corr:.4f}')
print(f'P-value:    {study_p:.6f}')
print('Result:', 'Statistically significant' if study_p < ALPHA else 'Not statistically significant')
```

## 2. Absences vs Final Grade — Pearson Correlation

**H₀:** There is no linear correlation between absences and final grade.

**H₁:** There is a linear correlation between absences and final grade.

```python
absence_corr, absence_p = stats.pearsonr(df['absences'], df['G3'])

print(f'Correlation: {absence_corr:.4f}')
print(f'P-value:    {absence_p:.6f}')
print('Result:', 'Statistically significant' if absence_p < ALPHA else 'Not statistically significant')
```

## 3. Previous Failures vs Final Grade — Pearson Correlation

**H₀:** There is no linear correlation between previous failures and final grade.

**H₁:** There is a linear correlation between previous failures and final grade.

```python
failure_corr, failure_p = stats.pearsonr(df['failures'], df['G3'])

print(f'Correlation: {failure_corr:.4f}')
print(f'P-value:    {failure_p:.6f}')
print('Result:', 'Statistically significant' if failure_p < ALPHA else 'Not statistically significant')
```

## 4. Gender vs Final Grade — Welch's Independent t-test

**H₀:** The mean final grade is the same across the two gender groups.

**H₁:** The mean final grade differs between the two gender groups.

Welch's t-test is used because it does not require equal population variances.

```python
male_scores = df.loc[df['sex'] == 'M', 'G3'].dropna()
female_scores = df.loc[df['sex'] == 'F', 'G3'].dropna()

gender_t, gender_p = stats.ttest_ind(
    male_scores,
    female_scores,
    equal_var=False
)

print(f'Male mean:   {male_scores.mean():.2f}')
print(f'Female mean: {female_scores.mean():.2f}')
print(f'T-statistic: {gender_t:.4f}')
print(f'P-value:     {gender_p:.6f}')
print('Result:', 'Statistically significant difference' if gender_p < ALPHA else 'No statistically significant difference detected')
```

## 5. Internet Access vs Final Grade — Welch's Independent t-test

**H₀:** Mean final grades are the same for students with and without internet access.

**H₁:** Mean final grades differ between the groups.

```python
internet_yes = df.loc[df['internet'] == 'yes', 'G3'].dropna()
internet_no = df.loc[df['internet'] == 'no', 'G3'].dropna()

internet_t, internet_p = stats.ttest_ind(
    internet_yes,
    internet_no,
    equal_var=False
)

print(f'Internet = Yes mean: {internet_yes.mean():.2f}')
print(f'Internet = No mean:  {internet_no.mean():.2f}')
print(f'T-statistic:          {internet_t:.4f}')
print(f'P-value:              {internet_p:.6f}')
print('Result:', 'Statistically significant difference' if internet_p < ALPHA else 'No statistically significant difference detected')
```

## 6. School vs Final Grade — Welch's Independent t-test

**H₀:** Mean final grades are the same across the two schools.

**H₁:** Mean final grades differ between the two schools.

```python
school_gp = df.loc[df['school'] == 'GP', 'G3'].dropna()
school_ms = df.loc[df['school'] == 'MS', 'G3'].dropna()

school_t, school_p = stats.ttest_ind(
    school_gp,
    school_ms,
    equal_var=False
)

print(f'GP mean:       {school_gp.mean():.2f}')
print(f'MS mean:       {school_ms.mean():.2f}')
print(f'T-statistic:   {school_t:.4f}')
print(f'P-value:       {school_p:.6f}')
print('Result:', 'Statistically significant difference' if school_p < ALPHA else 'No statistically significant difference detected')
```

## 7. Study Time vs Final Grade — One-way ANOVA

There are four study-time groups, so we use one-way ANOVA rather than repeatedly running t-tests.

**H₀:** All study-time groups have the same mean final grade.

**H₁:** At least one study-time group has a different mean final grade.

```python
study_groups = [
    group['G3'].dropna().values
    for _, group in df.groupby('studytime')
]

anova_f, anova_p = stats.f_oneway(*study_groups)

print(f'F-statistic: {anova_f:.4f}')
print(f'P-value:     {anova_p:.6f}')
print('Result:', 'At least one group differs significantly' if anova_p < ALPHA else 'No statistically significant difference detected')

print('\nMean final grade by study-time group:')
display(df.groupby('studytime')['G3'].agg(['count', 'mean', 'median']).round(2))
```

## 8. G1 and G2 vs G3 — Correlation Analysis

`G3` is the final grade, while `G1` and `G2` are earlier-period grades.

These variables are especially important for the later ML stage because using information that is only available after the prediction point can create **target leakage**.

```python
g_grade_results = []

for col in ['G1', 'G2']:
    corr, p = stats.pearsonr(df[col], df['G3'])
    g_grade_results.append({
        'Variable': col,
        'Correlation': corr,
        'P-value': p,
        'Significant': 'Yes' if p < ALPHA else 'No'
    })

g_grade_df = pd.DataFrame(g_grade_results)
display(g_grade_df.round(4))
```

## 9. Statistical Test Summary

This table collects the main results in one place.

```python
summary = pd.DataFrame([
    ['Study Time vs G3', 'Pearson correlation', study_corr, study_p],
    ['Absences vs G3', 'Pearson correlation', absence_corr, absence_p],
    ['Previous Failures vs G3', 'Pearson correlation', failure_corr, failure_p],
    ['Gender vs G3', "Welch's t-test", gender_t, gender_p],
    ['Internet Access vs G3', "Welch's t-test", internet_t, internet_p],
    ['School vs G3', "Welch's t-test", school_t, school_p],
    ['Study Time Groups vs G3', 'One-way ANOVA', anova_f, anova_p]
], columns=['Analysis', 'Test', 'Statistic', 'P-value'])

summary['Significant (α=0.05)'] = np.where(
    summary['P-value'] < ALPHA, 'Yes', 'No'
)

display(summary.round(4))
```

## 10. Automatically Generate Interpretation Notes

These are factual statistical interpretations of the tests. Statistical significance does **not** establish causation.

```python
print('STATISTICAL INTERPRETATION')
print('=' * 70)

tests = [
    ('Study time vs final grade', study_p),
    ('Absences vs final grade', absence_p),
    ('Previous failures vs final grade', failure_p),
    ('Gender group difference', gender_p),
    ('Internet access group difference', internet_p),
    ('School group difference', school_p),
    ('Study-time group difference', anova_p)
]

for name, p in tests:
    if p < ALPHA:
        print(f'✓ {name}: statistically significant (p={p:.6f})')
    else:
        print(f'• {name}: not statistically significant (p={p:.6f})')

print('\nNote: Statistical association/difference does not prove causation.')
```

This analysis provides statistically tested evidence about relationships and group differences in student performance.
