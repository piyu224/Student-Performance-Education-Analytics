```python
#1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
#2. Load the dataset
df = pd.read_csv("../data/student-mat.csv", sep=";")

df.head()
```

```python
#3. Check dataset dimensions
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
#4. Understand all columns
df.columns.tolist()
```

```python
#5. Check data types
df.info()
```

```python
#6. Check missing values
missing = df.isnull().sum()

missing[missing > 0]
```

```python
#7. Check duplicate records
print("Duplicate rows:", df.duplicated().sum())
```

```python
#8. Separate numerical and categorical columns
numerical_cols = df.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_cols = df.select_dtypes(
    include=["object"]
).columns

print("Numerical columns:")
print(numerical_cols.tolist())

print("\nCategorical columns:")
print(categorical_cols.tolist())
```

```python
#9. Check unique values
for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
#10. Check numerical distributions
df[numerical_cols].describe().T
```

```python
#11. Check our target variable
print(df["G3"].value_counts().sort_index())
```

```python
print("Average final grade:", df["G3"].mean())
print("Median final grade:", df["G3"].median())
print("Minimum:", df["G3"].min())
print("Maximum:", df["G3"].max())
```

```python
#12. Create a performance category
def performance_category(score):
    if score < 10:
        return "Fail"
    elif score < 14:
        return "Average"
    elif score < 17:
        return "Good"
    else:
        return "Excellent"

df["performance_category"] = df["G3"].apply(performance_category)
```

```python
df["performance_category"].value_counts()
```

```python
#13. Create a pass/fail column
df["pass"] = np.where(df["G3"] >= 10, "Pass", "Fail")
```

```python
df["pass"].value_counts()
```

```python
pass_rate = (df["pass"] == "Pass").mean() * 100
fail_rate = (df["pass"] == "Fail").mean() * 100

print(f"Pass Rate: {pass_rate:.2f}%")
print(f"Fail Rate: {fail_rate:.2f}%")
```
